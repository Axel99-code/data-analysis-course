### 4) Limpieza: aplicar las decisiones

Ahora sí, implementen las decisiones del Paso 3. Recuerden:

- Después manejen nulos
- Después conviertan tipos

```python
# Limpieza de pacientes
df_pacientes_limpio = df_pacientes.copy()

#duplicados

pacientes_sin_duplicados = df_pacientes.drop_duplicates(["nombre","dni","obra_social","fecha_nacimiento","ciudad","obra_social"])
print(f"Filas originales: {len(df_pacientes)}")
print(f"Sin duplicados exactos: {len(pacientes_sin_duplicados)}")

#nulos

df_pacientes_limpio["dni"] = df_pacientes_limpio["dni"].fillna("Sin datos")

df_pacientes_limpio["fecha_nacimiento"] = df_pacientes_limpio["fecha_nacimiento"].fillna("Sin datos")

df_pacientes_limpio["ciudad"] = df_pacientes_limpio["ciudad"].fillna("Sin recidencia")

df_pacientes_limpio["obra_social"] = df_pacientes_limpio["obra_social"].fillna("Sin datos")

#tipo de dato

df_pacientes_limpio['fecha_nacimiento'] = pd.to_datetime(df_pacientes_limpio['fecha_nacimiento'], errors='coerce')

print(df_pacientes_limpio["fecha_nacimiento"].dtypes)

print(len(df_pacientes_limpio))

# Limpieza de médicos

df_medicos_limpio = df_medicos.copy()

#duplicados

df_medicos_limpio = df_medicos.drop_duplicates(['nombre','especialidad','matricula'])
print(f"Filas originales: {len(df_medicos)}")
print(f"Sin duplicados exactos: {len(df_medicos_limpio)}")

#nulos

df_medicos_limpio["activo"] = df_medicos_limpio["activo"].fillna("Sin datos")

df_medicos_limpio["matricula"] = df_medicos_limpio["matricula"].fillna("Sin datos")

# Limpieza de turnos
df_turnos_limpio = df_turnos.copy()
#duplicados
turnos_sin_duplicados = df_turnos.drop_duplicates(["paciente_id","medico_id","fecha","hora","motivo","estado"])
print(f"Filas originales: {len(df_turnos)}")
print(f"Sin duplicados exactos: {len(turnos_sin_duplicados)}")
#nulos
df_turnos_limpio["hora"] = df_turnos_limpio["hora"].fillna("Sin datos")

df_turnos_limpio["motivo"] = df_turnos_limpio["motivo"].fillna("Sin datos")

df_turnos_limpio["costo"] = df_turnos_limpio["costo"].fillna("Sin datos")

df_turnos_limpio["costo"] = pd.to_numeric(df_turnos_limpio["costo"], errors='coerce')

df_turnos_limpio = df_turnos_limpio.dropna(subset=["costo"])

df_turnos_limpio = df_turnos_limpio[df_turnos_limpio["costo"] >= 0]

df_turnos_limpio['medico_id'] = df_turnos_limpio['medico_id'].replace(6, 1)

#tipo de dato

df_turnos_limpio['fecha'] = pd.to_datetime(df_turnos_limpio['fecha'], errors='coerce')

df_turnos_limpio["costo"] = df_turnos_limpio["costo"].astype(float)

print(df_turnos_limpio["fecha"].dtypes)

print(len(df_turnos_limpio))
```

> **Por qué `.copy()`:** cuando hacen `df_pacientes_limpio = df_pacientes`, no están creando una copia — están apuntando al mismo objeto. Cualquier cambio en `df_pacientes_limpio` modifica también `df_pacientes`. `.copy()` crea una copia independiente, así conservan los datos originales para comparar.
