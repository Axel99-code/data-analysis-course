```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

VENTAS MENSUALES

```python
#1.1
ventas = pd.read_csv("ventas_mensuales.csv", parse_dates=["mes"])

fig, ax = plt.subplots()
ax.bar(ventas["mes"].dt.strftime("%b"), ventas["ingresos"])
plt.show()
```

```python
#1.2
colores = ["#B0B0B0"] * len(ventas)
mes_pico = ventas["ingresos"].idxmax()
colores[mes_pico] = "#2E86AB"

ventas = pd.read_csv("ventas_mensuales.csv", parse_dates=["mes"])

fig, ax = plt.subplots()
ax.bar(ventas["mes"].dt.strftime("%b"), ventas["ingresos"], color=colores)

ax.set_title("Diciembre concentró el 22% de los ingresos del año", loc="left", fontsize=13, weight="bold")
ax.set_ylabel("Ingresos (ARS)")
ax.spines[["top", "right"]].set_visible(False)  # quitamos los bordes que no aportan información
ax.yaxis.grid(True, alpha=0.3)
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```

```python
#1.3
sns.set_style("whitegrid")

fig, ax = plt.subplots(figsize=(9, 5))

sns.barplot(
    data=ventas,
    x=ventas["mes"].dt.strftime("%b"),
    y="ingresos",
    color = "#B0B0B0",
    ax=ax,
)
ax.patches[mes_pico].set_facecolor("#2E86AB")  # resaltamos la barra del mes pico
ax.set_title("Diciembre concentró el 22% de los ingresos del año", loc="left", weight="bold")
```

```python
#1.4
colores = ["#B0B0B0"] * len(ventas)
mes_pico = ventas["ingresos"].idxmax()
colores[mes_pico] = "#2E86AB"
ventas = pd.read_csv("ventas_mensuales.csv", parse_dates=["mes"])

fig, ax = plt.subplots()
ax.bar(ventas["mes"].dt.strftime("%b"), ventas["ingresos"], color=colores)

ax.set_title("Diciembre concentró el 22% de los ingresos del año", loc="left", fontsize=13, weight="bold")
ax.set_ylabel("Ingresos (ARS)")
ax.spines[["top", "right"]].set_visible(False)  # quitamos los bordes que no aportan información
ax.yaxis.grid(True, alpha=0.3)
ax.set_axisbelow(True)

valor_pico = ventas["ingresos"].iloc[mes_pico]
ax.annotate(
    "Mes con mayores\ningresos",     # Texto de la anotación
    xy=(mes_pico, valor_pico),       # Punta de la flecha: X = número de barra, Y = altura de ingresos
    xytext=(mes_pico + 1.2, valor_pico * 0.9), # Posición del texto (un poco a la derecha y abajo)
    arrowprops=dict(
        facecolor="#2E86AB",         # Mismo color azul de tu barra destacada
        edgecolor="none",            # Sin borde para la flecha
        shrink=0.08,                 # Espacio para que no toque exactamente la barra ni el texto
        width=1.5,                   # Grosor del cuerpo de la flecha
        headwidth=7                  # Ancho de la punta de la flecha
    ),
    fontsize=10,
    color="#2E86AB",                 # Color del texto igual al de la barra destacada
    weight="bold"                    # Texto en negrita para resaltar
)

plt.tight_layout()
plt.show()
```

```python
#2.1
# 1. Cargamos el nuevo archivo CSV
ventas = pd.read_csv("ventas_por_categoria.csv")

# 2. Creamos la figura con un tamaño cuadrado para que la torta no se deforme
fig, ax = plt.subplots(figsize=(6, 6))

# 3. Generamos un color diferente para cada categoría
# 'tab10' es excelente para diferenciar categorías individuales (ropa, tecnología, comida, etc.)
cmap = plt.get_cmap("tab10")
colores = [cmap(i) for i in range(len(ventas))]

# 4. Dibujamos el gráfico de torta
ax.pie(
    ventas["ingresos"],                       # Valores numéricos
    labels=ventas["categoria"],              # Nombres de las categorías
    colors=colores,                          # Lista de colores distintos
    autopct="%1.0f%%",                       # Muestra el porcentaje sin decimales
    startangle=90,                           # Comienza a dibujar desde arriba
    pctdistance=0.75,                        # Posición de los números del porcentaje
    textprops={"fontsize": 11}               # Tamaño de la letra
)

# 5. Título del gráfico
ax.set_title("Distribución de Ingresos por Categoría", loc="center", fontsize=14, weight="bold")

plt.tight_layout()
plt.show()
```
